```python
import pandas as pd
import numpy as np
```

```python
file = r'data\medicine_dataset.csv'
data = pd.read_csv(file)
df = pd.DataFrame(data)
df.head()
```

```python
df = df.drop('id', axis = 1)
df.head()
```

```python
print("Data Shape: ", df.shape)
```

```python
initial_features = df.columns
print("Initial feature: ", initial_features)
```

```python
df['Substitutes'] = df[['substitute0', 'substitute1', 'substitute2', 'substitute3', 'substitute4']] \
                    .astype(str)\
                    .agg(', '.join, axis=1)

df['Side Effects'] = df[['sideEffect0', 'sideEffect1', 'sideEffect2',
       'sideEffect3', 'sideEffect4', 'sideEffect5', 'sideEffect6',
       'sideEffect7', 'sideEffect8', 'sideEffect9', 'sideEffect10',
       'sideEffect11', 'sideEffect12', 'sideEffect13', 'sideEffect14',
       'sideEffect15', 'sideEffect16', 'sideEffect17', 'sideEffect18',
       'sideEffect19', 'sideEffect20', 'sideEffect21', 'sideEffect22',
       'sideEffect23', 'sideEffect24', 'sideEffect25', 'sideEffect26',
       'sideEffect27', 'sideEffect28', 'sideEffect29', 'sideEffect30',
       'sideEffect31', 'sideEffect32', 'sideEffect33', 'sideEffect34',
       'sideEffect35', 'sideEffect36', 'sideEffect37', 'sideEffect38',
       'sideEffect39', 'sideEffect40', 'sideEffect41']].astype(str).agg(', '.join, axis = 1)

df['Uses'] = df[['use0', 'use1', 'use2', 'use3', 'use4']].astype(str).agg(', '.join, axis = 1)
```

```python
df.head()
```

```python
keep_cols = ['name', 'Chemical Class', 'Habit Forming', 'Therapeutic Class', 'Action Class']

features_to_drop = [cols for cols in initial_features.tolist() if cols not in keep_cols]
```

```python
df = df.drop(columns = features_to_drop, axis = 1)
df.head()
```

```python
df.replace(np.nan, '', inplace = True)
df.head()
```

```python
df['Substitutes'] = df['Substitutes'].str.replace('nan', '', regex = False).str.replace(r'(,\s*)+', ', ', regex = True).str.strip(', ')
df['Side Effects'] = df['Side Effects'].str.replace('nan', '', regex = False).str.replace(r'(,\s*)+', ', ', regex = True).str.strip(', ')
df['Uses'] = df['Uses'].str.replace('nan', '', regex = False).str.replace(r'(,\s*)+', ', ', regex = True).str.strip(', ')
df.head()
```

```python
df.to_csv("./data/Preprocessed Medicine Dataset.csv", index = False)
```
